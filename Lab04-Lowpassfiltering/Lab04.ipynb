1)	Aplique o filtro de Média na imagem abaixo para diferentes tamanhos de máscaras.
- 3x3; 7x7; 3x3 aplicado três vezes sequenciais.

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
# code here
img1 = cv2.imread("sta2.png", cv2.IMREAD_GRAYSCALE)

# Filtro média 3x3
media_3x3 = cv2.blur(img1, (3, 3))

# Filtro média 7x7
media_7x7 = cv2.blur(img1, (7, 7))

# Filtro média 3x3 aplicado três vezes
media_3x3_1 = cv2.blur(img1, (3, 3))
media_3x3_2 = cv2.blur(media_3x3_1, (3, 3))
media_3x3_3 = cv2.blur(media_3x3_2, (3, 3))

# Exibição
plt.figure(figsize=(12,8))

plt.subplot(2,2,1)
plt.imshow(img1, cmap='gray', vmin=0, vmax=255)
plt.title('Original')
plt.axis('off')

plt.subplot(2,2,2)
plt.imshow(media_3x3, cmap='gray', vmin=0, vmax=255)
plt.title('Média 3x3')
plt.axis('off')

plt.subplot(2,2,3)
plt.imshow(media_7x7, cmap='gray', vmin=0, vmax=255)
plt.title('Média 7x7')
plt.axis('off')

plt.subplot(2,2,4)
plt.imshow(media_3x3_3, cmap='gray', vmin=0, vmax=255)
plt.title('Média 3x3 (3 vezes)')
plt.axis('off')

plt.tight_layout()
plt.show()
```

2)	Aplique o filtro Gaussiano à imagem abaixo para diferentes valores de desvio padrão (σ) e tamanho de máscara.
- σ = 1.0 (kernel size 5×5).	σ = 2.0 (kernel size 9×9).	σ =  4.0 (kernel size 15×15).

```python
img2 = cv2.imread("ben2.png", cv2.IMREAD_GRAYSCALE)


gauss_1 = cv2.GaussianBlur(img2, (5, 5), sigmaX=1.0)
gauss_2 = cv2.GaussianBlur(img2, (9, 9), sigmaX=2.0)
gauss_3 = cv2.GaussianBlur(img2, (15, 15), sigmaX=4.0)


plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(img2, cmap='gray', vmin=0, vmax=255)
plt.title('Original')
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(gauss_1, cmap='gray', vmin=0, vmax=255)
plt.title('sigma=1.0 (5x5)')
plt.axis('off')

plt.subplot(2, 2, 3)
plt.imshow(gauss_2, cmap='gray', vmin=0, vmax=255)
plt.title('sigma=2.0 (9x9)')
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(gauss_3, cmap='gray', vmin=0, vmax=255)
plt.title('sigma=4.0 (15x15)')
plt.axis('off')

plt.tight_layout()
plt.show()
```

3)	Aplique o filtro de Mediana para a imagem abaixo, que apresenta um ruído *salt and pepper*

4)	Aplique o filtro de Mediana para a imagem abaixo, que apresenta um ruído salt and pepper. Não utilizar a função pronta do openvc, mas pode usar numpy.median()
- 3×3 filter;	7×7 filter

```python
img3 = cv2.imread("sta2noi1.png", cv2.IMREAD_GRAYSCALE)

def filtro_mediana(img, tamanho):
    margem = tamanho // 2

    #adicionar bordas
    img_pad = np.pad(img, margem, mode='edge')

    resultado = np.zeros_like(img)

    for i in range(img.shape[0]):
        for j in range(img.shape[1]):
            janela = img_pad[i:i+tamanho, j:j+tamanho]
            resultado[i, j] = np.median(janela)

    return resultado


mediana_3x3 = filtro_mediana(img3, 3)
mediana_7x7 = filtro_mediana(img3, 7)


plt.figure(figsize=(12,6))

plt.subplot(1,3,1)
plt.imshow(img3, cmap='gray', vmin=0, vmax=255)
plt.title('Original')
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(mediana_3x3, cmap='gray', vmin=0, vmax=255)
plt.title('Mediana 3x3')
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(mediana_7x7, cmap='gray', vmin=0, vmax=255)
plt.title('Mediana 7x7')
plt.axis('off')

plt.tight_layout()
plt.show()
```
